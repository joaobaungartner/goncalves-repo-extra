```python
import pandas as pd
import numpy as np

# =========================================================
# ROTEIRO COMPLETO (EM CÓDIGO) PARA TESTAR "MISSINGNESS"
# Para cada coluna com NaN (target), testa associação do NaN
# com TODAS as outras features (numéricas e categóricas).
# =========================================================

# 1) Identifica colunas com faltantes
missing_cols = df.columns[df.isna().any()].tolist()
print("Colunas com valores ausentes:", missing_cols)

# 2) Define quais colunas são categóricas vs numéricas
# (heurística: object/category => categórica; resto => numérica)
cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()
num_cols = [c for c in df.columns if c not in cat_cols]

# 3) Funções auxiliares
def cramers_v(confusion_matrix: pd.DataFrame) -> float:
    """Cramér's V para associação entre variáveis categóricas."""
    n = confusion_matrix.values.sum()
    if n == 0:
        return np.nan

    observed = confusion_matrix.values
    row_sum = observed.sum(axis=1, keepdims=True)
    col_sum = observed.sum(axis=0, keepdims=True)
    expected = row_sum @ col_sum / n

    # evita divisão por zero
    with np.errstate(divide="ignore", invalid="ignore"):
        chi2 = np.nansum((observed - expected) ** 2 / expected)

    r, k = observed.shape
    if min(r, k) <= 1:
        return 0.0

    return np.sqrt((chi2 / n) / (min(r - 1, k - 1)))


def point_biserial_like(flag: pd.Series, x: pd.Series) -> float:
    """
    Medida simples (correlação de Pearson entre flag 0/1 e variável numérica).
    Não é teste estatístico formal, mas é uma métrica útil.
    """
    tmp = pd.DataFrame({"flag": flag.astype(int), "x": x})
    tmp = tmp.dropna()
    if tmp["x"].nunique() <= 1 or tmp["flag"].nunique() <= 1:
        return np.nan
    return tmp["flag"].corr(tmp["x"])


def effect_size_mean_diff(flag: pd.Series, x: pd.Series) -> float:
    """
    Efeito de diferença de médias padronizada (Cohen's d) entre:
    grupo flag=1 (missing) e flag=0 (não missing)
    """
    tmp = pd.DataFrame({"flag": flag.astype(int), "x": x}).dropna()
    if tmp["flag"].nunique() < 2:
        return np.nan

    g0 = tmp[tmp["flag"] == 0]["x"]
    g1 = tmp[tmp["flag"] == 1]["x"]
    if len(g0) < 2 or len(g1) < 2:
        return np.nan

    m0, m1 = g0.mean(), g1.mean()
    s0, s1 = g0.std(ddof=1), g1.std(ddof=1)
    sp = np.sqrt(((len(g0) - 1) * s0**2 + (len(g1) - 1) * s1**2) / (len(g0) + len(g1) - 2))
    if sp == 0 or np.isnan(sp):
        return np.nan
    return (m1 - m0) / sp


# 4) Loop principal: para cada coluna com NaN (target),
#    testa associação do missing com TODAS as outras colunas.
resultados = []

for target in missing_cols:
    flag = df[target].isna()  # True se está ausente (missing)
    missing_rate = flag.mean() * 100

    for feat in df.columns:
        if feat == target:
            continue

        # Caso a feature seja categórica
        if feat in cat_cols:
            ct = pd.crosstab(flag, df[feat], dropna=False)
            v = cramers_v(ct)
            resultados.append({
                "target_missing_col": target,
                "feature_testada": feat,
                "tipo_feature": "categorica",
                "missing_%_target": round(missing_rate, 2),
                "metrica": "cramers_v",
                "valor_metrica": v
            })

        # Caso a feature seja numérica
        else:
            r = point_biserial_like(flag, df[feat])
            d = effect_size_mean_diff(flag, df[feat])
            resultados.append({
                "target_missing_col": target,
                "feature_testada": feat,
                "tipo_feature": "numerica",
                "missing_%_target": round(missing_rate, 2),
                "metrica": "corr_flag_num (pearson)",
                "valor_metrica": r
            })
            resultados.append({
                "target_missing_col": target,
                "feature_testada": feat,
                "tipo_feature": "numerica",
                "missing_%_target": round(missing_rate, 2),
                "metrica": "cohen_d (diff medias)",
                "valor_metrica": d
            })

# 5) DataFrame final de resultados (ranking por magnitude)
res_df = pd.DataFrame(resultados)

# Cria coluna auxiliar para ordenar por "força" (magnitude)
res_df["abs_valor_metrica"] = res_df["valor_metrica"].abs()

# Ordena do mais associado para o menos associado
res_df_sorted = res_df.sort_values(["target_missing_col", "abs_valor_metrica"], ascending=[True, False])

display(res_df_sorted)

# 6) (Opcional) Top N relações mais fortes por target
TOP_N = 10
top_por_target = (
    res_df_sorted
    .groupby("target_missing_col")
    .head(TOP_N)
    .drop(columns=["abs_valor_metrica"])
)

print(f"\nTOP {TOP_N} associações por coluna com NaN:")
display(top_por_target)
```

```python
import pandas as pd
import numpy as np

from scipy.stats import chi2, ttest_ind, chi2_contingency
import numpy.linalg as la

# =========================
# 1) Carregar base
# =========================
path = "BASE_1_Manga_Produtos.xlsx"  # ajuste se necessário
df = pd.read_excel(path)

print("shape:", df.shape)

# =========================
# 2) Diagnóstico de missing
# =========================
missing_counts = df.isna().sum().sort_values(ascending=False)
missing_pct = (missing_counts / len(df) * 100).round(2)

miss_summary = pd.DataFrame({
    "missing_count": missing_counts,
    "missing_pct": missing_pct
})

print("\nResumo de missing (apenas colunas com NA):")
print(miss_summary[miss_summary["missing_count"] > 0])

# (opcional) Visual rápido do padrão de missing
try:
    import missingno as msno
    msno.matrix(df)
    msno.heatmap(df)
except Exception as e:
    print("\n[Opcional] Para gráficos de missing, instale: pip install missingno")
    print("Motivo:", e)

# =========================
# 3) Little’s MCAR test (para numéricos)
# =========================
def little_mcar_test(data: pd.DataFrame, ridge: float = 1e-6):
    """
    Versão prática do Little’s MCAR test para variáveis numéricas.
    Interpretação:
      - p > 0.05: não rejeita MCAR (compatível com MCAR)
      - p <= 0.05: rejeita MCAR (então é MAR ou MNAR)
    """
    X = data.copy()
    mask = ~X.isna()

    # padrão de missing por linha (tuple de True/False)
    patterns = mask.apply(lambda row: tuple(row.values.tolist()), axis=1)

    # médias gerais (por coluna)
    mu = X.mean(skipna=True).values

    # covariância pairwise (aproximação útil na prática)
    S = (X - X.mean(skipna=True)).cov(min_periods=2).values
    S = S + np.eye(S.shape[0]) * ridge  # regularização p/ inversão

    stat = 0.0
    df_total = 0

    unique_patterns = patterns.value_counts().to_dict()

    for pat_tuple, n in unique_patterns.items():
        pat = np.array(pat_tuple, dtype=bool)
        obs_idx = np.where(pat)[0]
        m_p = len(obs_idx)

        if m_p == 0:
            continue

        rows = patterns == pat_tuple
        cols_obs = X.columns[obs_idx]

        X_p = X.loc[rows, cols_obs]
        mu_p = X_p.mean(skipna=True).values
        mu_obs = mu[obs_idx]

        diff = (mu_p - mu_obs).reshape(-1, 1)
        S_obs = S[np.ix_(obs_idx, obs_idx)]

        try:
            inv = la.inv(S_obs)
        except la.LinAlgError:
            inv = la.pinv(S_obs)

        stat += n * float(diff.T @ inv @ diff)
        df_total += m_p

    df_chi = df_total - X.shape[1]
    p_value = 1 - chi2.cdf(stat, df_chi) if df_chi > 0 else np.nan

    return {"chi2": stat, "df": df_chi, "p_value": p_value, "n_patterns": len(unique_patterns)}

num_cols = df.select_dtypes(include=[np.number]).columns.tolist()

if len(num_cols) >= 2:
    little_res = little_mcar_test(df[num_cols])
    print("\nLittle’s MCAR test (numéricos):", little_res)
else:
    print("\nPoucas colunas numéricas para Little’s test.")

# =========================
# 4) Testes de associação (missing indicator vs variáveis observadas)
# =========================
def missingness_association(df: pd.DataFrame, target_col: str, alpha=0.05):
    """
    Cria indicador: target_col_is_missing.
    Testa se esse indicador se associa com cada outra variável:
      - preditor numérico: t-test (Welch)
      - preditor categórico: chi-square
    Aplica Bonferroni para múltiplos testes.
    """
    y = df[target_col].isna().astype(int)  # 1 = missing

    results = []
    for col in df.columns:
        if col == target_col:
            continue

        x = df[col]
        if x.isna().all():
            continue

        if pd.api.types.is_numeric_dtype(x):
            g0 = x[y == 0].dropna()
            g1 = x[y == 1].dropna()
            if len(g0) < 3 or len(g1) < 3:
                continue
            stat, p = ttest_ind(g0, g1, equal_var=False)
            results.append((col, "ttest", p, float(g0.mean()), float(g1.mean()), len(g0), len(g1)))
        else:
            tab = pd.crosstab(y, x)  # mantém NA do preditor fora, ok
            if tab.shape[1] < 2:
                continue
            chi, p, _, _ = chi2_contingency(tab)
            results.append((col, "chi2", p, None, None, int(tab.loc[0].sum()), int(tab.loc[1].sum())))

    res = pd.DataFrame(
        results,
        columns=["predictor", "test", "p_value", "mean_nonmiss", "mean_miss", "n_nonmiss", "n_miss"]
    ).sort_values("p_value")

    if len(res) > 0:
        res["p_bonf"] = np.minimum(res["p_value"] * len(res), 1.0)
        res["sig_bonf"] = res["p_bonf"] < alpha

    return res

# Varre todas as colunas com missing e guarda os achados relevantes
findings = {}
for col in df.columns:
    if df[col].isna().any():
        res = missingness_association(df, col)
        sig = res[res.get("sig_bonf", False) == True] if len(res) else res
        if len(sig) > 0:
            findings[col] = sig[["predictor", "test", "p_value", "p_bonf"]].head(10)

print("\nColunas com evidência de NÃO-MCAR (associação significativa após Bonferroni):")
if len(findings) == 0:
    print("Nenhuma encontrada (isso favorece MCAR, mas não prova).")
else:
    for k, v in findings.items():
        print(f"\n>> {k}")
        print(v.to_string(index=False))
```
